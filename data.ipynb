```python
import pandas as pd
import subprocess
import os

VIDEO_ROOT = "./data/EPIC-KITCHENS"
OUTPUT_ROOT = "./data/clips"
ANNOTATIONS_FILE = "./epic-kitchens-55-annotations/EPIC_train_action_labels.csv"

TARGET_VIDEOS = ["P02_05", "P14_03"]

def slice_clips():
    df = pd.read_csv(ANNOTATIONS_FILE)
    
    df = df[df['video_id'].isin(TARGET_VIDEOS)]
    print(len(df), "actions")

    os.makedirs(OUTPUT_ROOT, exist_ok=True)

    count = 0
    for idx, row in df.iterrows():
        vid_id = row['video_id']
        start_time = row['start_timestamp']
        stop_time = row['stop_timestamp']
        narration = row['narration'].replace(" ", "_")
        
        input_video_path = os.path.join(VIDEO_ROOT, vid_id.split('_')[0], "./videos", f"{vid_id}.MP4")
        
        output_clip_name = f"{vid_id}_{row['uid']}_{narration}.mp4"
        output_clip_path = os.path.join(OUTPUT_ROOT, output_clip_name)

        if not os.path.exists(input_video_path):
            print("tf")
            continue

        cmd = [
            "ffmpeg", "-y",
            "-ss", start_time,
            "-to", stop_time,
            "-i", input_video_path,
            "-c", "copy",
            "-loglevel", "error",
            output_clip_path
        ]
        
        subprocess.run(cmd)
        count += 1
        print(f"Extracted {count}/{len(df)}: {narration}")

    print(f"\n\n{count} clips extracted to {OUTPUT_ROOT}")


```

```python
slice_clips()
```
